## 2. 클래스 정의 및 인스턴스 생성

```python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year
        self.is_running = False

    def start_engine(self):
        self.is_running = True
        print(f"{self.brand} {self.model} 엔진이 시작되었습니다.")

    def stop_engine(self):
        self.is_running = False
        print(f"{self.brand} {self.model} 엔진이 정지되었습니다.")

my_car = Car("Hyundai", "Sonata", 2020)
print(f"내 차: {my_car.brand} {my_car.model} ({my_car.year})")
my_car.start_engine()
print(f"내 차 엔진 상태: {my_car.is_running}")
```

### 2.6 객체의 문자열 표현 (`__str__`, `__repr__`)

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"Point({self.x}, {self.y})"

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"

p = Point(10, 20)
print(f"print(p): {p}")
print(f"repr(p): {repr(p)}")
```

### 2.7 `dataclasses` 모듈

```python
from dataclasses import dataclass, field
from typing import List

@dataclass
class Product:
    name: str
    price: float
    quantity: int = 0
    tags: List[str] = field(default_factory=list)

    def total_price(self) -> float:
        return self.price * self.quantity

p1 = Product("Laptop", 1200.0, 2)
print(p1)
print(f"{p1.name}의 총 가격: {p1.total_price()}원")
```
