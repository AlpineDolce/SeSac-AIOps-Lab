## 1. 타입 힌트 (Type Hinting)

```python
from typing import List, Dict, Optionaldef process_data(data: List[Dict[str, int]], name: Optional[str] = None) -> float:    """    데이터를 처리하고 평균 값을 반환합니다.    """    if name:        print(f"Processing data for {name}")    total = sum(item['value'] for item in data)    return total / len(data)sample_data = [{'value': 10}, {'value': 20}, {'value': 30}]average = process_data(sample_data, name="Test Data")print(f"평균 값: {average}")
```

## 2. 데코레이터 (Decorators)

```python
import timeimport functoolsdef timer(func):    @functools.wraps(func)    def wrapper(*args, **kwargs):        start_time = time.time()        result = func(*args, **kwargs)        end_time = time.time()        print(f'''{{func.__name__}}'' 함수 실행 시간: {end_time - start_time:.4f}초')        return result    return wrapper@timerdef process_large_data(size: int):    time.sleep(size * 0.1)process_large_data(3)
```

### 2.2.1 `functools.lru_cache`를 이용한 메모이제이션

```python
import functools@functools.lru_cache(maxsize=None)def fibonacci(n):    if n <= 1:        return n    return fibonacci(n-1) + fibonacci(n-2)print(f"피보나치(20): {fibonacci(20)}")
```

```python
import asyncioimport timeasync def fetch_data(delay: int, name: str) -> str:    print(f'''{{name}}'' 데이터 가져오기 시작...')    await asyncio.sleep(delay)    print(f'''{{name}}'' 데이터 가져오기 완료!')    return f"{name}의 데이터"async def main():    start_time = time.time()    task1 = asyncio.create_task(fetch_data(2, "사용자 정보"))    task2 = asyncio.create_task(fetch_data(3, "상품 목록"))    data1 = await task1    data2 = await task2    end_time = time.time()    print(f"총 실행 시간: {end_time - start_time:.2f}초")# asyncio.run(main()) # .py 파일에서 실행
```
