## 1. 딕셔너리 (Dictionary)

```python
person = "{"name": "Alice", "age": 30, "city": "New York"}"
print(f"person 딕셔너리: {person}")

print(f"Alice의 나이: {person["age"]}")

person["age"] = 31
print(f"수정된 Alice의 나이: {person["age"]}")

person["occupation"] = "Engineer"
print(f"직업 추가 후: {person}")
```

### 1.3 딕셔너리 주요 메서드

```python
my_dict = {"a": 1, "b": 2, "c": 3}

print(f"키 목록: {my_dict.keys()}")
print(f"값 목록: {my_dict.values()}")
print(f"아이템 목록: {my_dict.items()}")

print(f""d"의 값 (기본값 0): {my_dict.get("d", 0)}")

my_dict.update({"b": 20, "d": 4})
print(f"update 후: {my_dict}")
```

### 1.4 딕셔너리 컴프리헨션

```python
squares_dict = {x: x**2 for x in range(5) if x % 2 == 0}
print(f"짝수의 제곱 딕셔너리: {squares_dict}")
```

### 1.5.1 `defaultdict`

```python
from collections import defaultdict

word_groups = defaultdict(list)
words = ["apple", "banana", "apricot", "berry", "cat"]

for word in words:
    word_groups[word[0]].append(word)

print(f"첫 글자별 단어 그룹: {word_groups}")
```

### 1.5.2 `Counter`

```python
from collections import Counter

my_list = ["apple", "banana", "apple", "cherry", "banana", "apple"]
counts = Counter(my_list)
print(f"리스트 요소 카운트: {counts}")
print(f"가장 흔한 2가지 요소: {counts.most_common(2)}")
```

## 2. 집합 (Set)

```python
my_set = {1, 2, 3, 2, 1} # 중복 제거됨
print(f"my_set: {my_set}")

my_set.add(4)
print(f"add(4) 후: {my_set}")

my_set.remove(3)
print(f"remove(3) 후: {my_set}")
```

### 2.3 집합 연산

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

print(f"합집합: {set1.union(set2)}")
print(f"교집합: {set1.intersection(set2)}")
print(f"차집합 (set1 - set2): {set1.difference(set2)}")
print(f"대칭 차집합: {set1.symmetric_difference(set2)}")
```
