## 2. 문자열 (String)

```python
s = "Hello, Python!"

print(f"첫 번째 문자: {s[0]}")
print(f"마지막 문자: {s[-1]}")
print(f"0~4까지: {s[0:5]}")
print(f"2칸씩 건너뛰기: {s[::2]}")
print(f"역순으로: {s[::-1]}")
```

### 2.2 문자열 주요 메서드

```python
text = "  python is fun  "
words = "apple,banana,cherry"

word_list = words.split(',')
print(f"split 결과: {word_list}")

joined_str = " ".join(word_list)
print(f"join 결과: {joined_str}")

processed_text = text.strip().replace("fun", "powerful").upper()
print(f"처리된 텍스트: {processed_text}")
```

## 3. 리스트 (List)

```python
my_list = [10, 20, 30]
print(f"초기 리스트: {my_list}")

my_list.append(40)
print(f"append(40) 후: {my_list}")

my_list.insert(1, 15)
print(f"insert(1, 15) 후: {my_list}")

my_list.pop()
print(f"pop() 후: {my_list}")

my_list.remove(20)
print(f"remove(20) 후: {my_list}")
```

### 3.3 리스트 정렬 (`sort()` vs `sorted()`)

```python
numbers = [3, 1, 4, 1, 5, 9, 2]
new_sorted_list = sorted(numbers)
print(f"sorted() 후 원본: {numbers}")
print(f"sorted() 반환값: {new_sorted_list}")

numbers.sort()
print(f"sort() 후: {numbers}")
```

### 3.5 리스트 컴프리헨션 (List Comprehensions)

```python
squares = [x**2 for x in range(1, 6)]
print(f"제곱 리스트: {squares}")

even_numbers = [x for x in range(10) if x % 2 == 0]
print(f"짝수 리스트: {even_numbers}")
```

## 4. 튜플 (Tuple)

```python
packed_tuple = 1, 2, 3
print(f"패킹된 튜플: {packed_tuple}")

a, b, c = packed_tuple
print(f"언패킹 결과: a={a}, b={b}, c={c}")

x, y = 10, 20
x, y = y, x
print(f"교환 후: x={x}, y={y}")
```
