## 1. 모듈 (Module)
모듈 실습을 위해 `my_module.py` 파일을 생성했다고 가정합니다.

````python
# my_module.py
PI = 3.14159

def add(a, b):
    return a + b

class MyClass:
    def __init__(self, name):
        self.name = name

    def greet(self):
        return f"Hello from MyClass, {self.name}!"

if __name__ == "__main__":
    print("my_module.py가 직접 실행되었습니다.")
````

### 1.1 모듈 생성 및 `import`

````python
# 이 노트북(또는 다른 .py 파일)에서 my_module을 임포트합니다.
import my_module

print(f"PI 값: {my_module.PI}")
print(f"덧셈 결과: {my_module.add(10, 5)}")
my_obj = my_module.MyClass("Alice")
print(my_obj.greet())
````

## 2. 패키지 (Package)
아래와 같은 디렉토리 구조를 가정하고 실습합니다.
```
my_project/
├── main.py
└── my_package/
    ├── __init__.py
    └── module_a.py
```

````python
# my_package/__init__.py
from .module_a import add_one

# my_package/module_a.py
def add_one(num):
    return num + 1
````

````python
# main.py 에서의 사용 예시
from my_package import add_one

print(f"패키지 레벨 호출: {add_one(10)}")
````
