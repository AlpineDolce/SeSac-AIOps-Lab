## 1. 단위 테스트 (Unit Testing) 와 `pytest`
실습을 위해 `my_functions.py`와 `test_my_functions.py`를 생성했다고 가정합니다.

```python
# my_functions.py
def add(a, b):
    return a + b

def divide(a, b):
    if b == 0:
        raise ValueError("0으로 나눌 수 없습니다.")
    return a / b
```

```python
# test_my_functions.py
from my_functions import add, divide
import pytest

def test_add():
    assert add(1, 2) == 3

def test_divide_by_zero():
    with pytest.raises(ValueError):
        divide(10, 0)

@pytest.mark.parametrize("a, b, expected", [
    (1, 2, 3), (-1, 1, 0), (0, 0, 0)
])
def test_add_parametrized(a, b, expected):
    assert add(a, b) == expected
```

## 4. 예외 처리 (Exception Handling)

```python
try:
    num1 = int(input("숫자를 입력하세요: "))
    num2 = int(input("다른 숫자를 입력하세요: "))
    result = num1 / num2
except ValueError:
    print("유효한 숫자가 아닙니다.")
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
else:
    print(f"나눗셈 결과: {result}")
finally:
    print("예외 처리 블록이 종료되었습니다.")
```

## 5. 정규표현식 (Regular Expressions)

```python
import re

text = "My phone number is 123-456-7890."

search_obj = re.search(r'\d+', text)
print(f"search('\d+'): {search_obj.group() if search_obj else 'No match'}")

findall_list = re.findall(r'\d+', text)
print(f"findall('\d+'): {findall_list}")

new_text = re.sub(r'\d{3}-\d{4}-\d{4}', '[PHONE_NUMBER]', text)
print(f"sub(): {new_text}")
```
