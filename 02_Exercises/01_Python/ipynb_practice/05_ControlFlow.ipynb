## 1. 조건문 (Conditional Statements)

```python
score = 75
if score >= 90:
    print("A 학점")
elif score >= 80:
    print("B 학점")
elif score >= 70:
    print("C 학점")
else:
    print("F 학점")
```

### 1.2 조건부 표현식 (Conditional Expression)

```python
score = 65
message = "합격" if score >= 60 else "불합격"
print(f"결과: {message}")
```

## 2. 반복문 (Loops)

### 2.1 `for` 루프: 시퀀스 순회

```python
fruits = ['apple', 'banana', 'cherry']for fruit in fruits:
    print(f"과일: {fruit}")

# 인덱스와 값 동시 순회 (enumerate)
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")
```

### 2.2 `while` 루프: 조건 기반 반복

```python
count = 1
while count <= 5:
    print(count)
    count += 1
```

## 3. 반복 제어

```python
# 1부터 10까지 출력하다가 5를 만나면 중단 (break)
for i in range(1, 11):
    if i == 5:
        print("5를 만났습니다. 반복 중단!")
        break
    print(i)

# 1부터 10까지 홀수만 출력 (continue)
for i in range(1, 11):
    if i % 2 == 0: # 짝수이면 건너뛰기
        continue
    print(i)
```

### 3.3 `else` 절을 갖는 반복문

```python
for i in range(5):
    print(i)
else:
    print("모든 반복을 완료했습니다.")
```
