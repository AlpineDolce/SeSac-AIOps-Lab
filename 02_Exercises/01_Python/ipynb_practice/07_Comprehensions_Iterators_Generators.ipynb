## 1. 컴프리헨션 (Comprehensions)

```python
# 리스트 컴프리헨션
squares = [x**2 for x in range(1, 6)]
print(f"제곱 리스트: {squares}")

# 딕셔너리 컴프리헨션
squares_dict = {x: x**2 for x in range(5) if x % 2 == 0}
print(f"짝수의 제곱 딕셔너리: {squares_dict}")

# 집합 컴프리헨션
unique_chars = {char for char in "hello world" if char.isalpha()}
print(f"고유한 문자 집합: {unique_chars}")
```

## 2. 이터레이터와 제너레이터

### 2.2 `yield` 키워드를 사용한 제너레이터 함수

```python
def count_up_to(max_val):
    count = 1
    while count <= max_val:
        yield count
        count += 1

counter = count_up_to(5)
print(f"제너레이터 타입: {type(counter)}")

for num in counter:
    print(num)
```

### 2.4 제너레이터 표현식 (Generator Expressions)

```python
gen_exp = (x * x for x in range(1000000))
print(f"제너레이터 표현식 타입: {type(gen_exp)}")

sum_of_squares = sum(x*x for x in range(10))
print(f"0부터 9까지 제곱의 합: {sum_of_squares}")
```

## 3. 유용한 내장 함수

```python
# zip()
names = ['Alice', 'Bob', 'Charlie']
ages = [25, 30, 35]
for name, age in zip(names, ages):
    print(f"이름: {name}, 나이: {age}")

# enumerate()
fruits = ['apple', 'banana', 'cherry']
for index, fruit in enumerate(fruits, start=1):
    print(f"{index}. {fruit}")

# all() and any()
numbers = [1, 2, 3, 4, 5]
print(f"모든 숫자가 0보다 큰가? {all(x > 0 for x in numbers)}")
print(f"짝수가 하나라도 있는가? {any(x % 2 == 0 for x in numbers)}")
```
