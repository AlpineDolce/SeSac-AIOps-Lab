## 2. 기본 자료형 (Primitive Data Types)

### 2.2.1 부동소수점 정밀도 문제와 `Decimal` 모듈 활용

```python
from decimal import Decimal, getcontext

# 기본 정밀도 설정 (28자리)
getcontext().prec = 28

# float의 부동소수점 오차
print(f"0.1 + 0.2 (float): {0.1 + 0.2}") # 0.30000000000000004

# Decimal을 사용한 정확한 계산
d1 = Decimal('0.1')
d2 = Decimal('0.2')
d3 = Decimal('0.3')
print(f"Decimal('0.1') + Decimal('0.2'): {d1 + d2}") # 0.3
print(f"d1 + d2 == d3: {d1 + d2 == d3}") # True

# 정밀도 제어 (반올림)
getcontext().prec = 4 # 정밀도를 4자리로 설정
d_pi = Decimal('3.1415926535')
print(f"Decimal(pi) with prec 4: {d_pi}") # 3.142 (자동 반올림)

# 특정 반올림 모드 적용
from decimal import ROUND_HALF_UP
getcontext().prec = 6
d_val = Decimal('2.5')
print(f"Decimal('2.5').quantize(Decimal('1.'), rounding=ROUND_HALF_UP): {d_val.quantize(Decimal('1.'), rounding=ROUND_HALF_UP)}") # 3
```

### 2.5.1 문자열 포매팅 (f-string, format(), %)

```python
name = "Alice"
age = 30
print(f"이름: {name}, 나이: {age}")
print(f"내년 나이: {age + 1}")
```

-   **문자열은 불변(immutable) 객체입니다.** 즉, 한 번 생성된 문자열은 내용을 변경할 수 없으며, 변경처럼 보이는 작업은 새로운 문자열을 생성하는 것입니다.
    -   **실무적 의미:** 문자열을 변경하는 것처럼 보이는 연산(예: `+`를 이용한 연결, `replace()` 메서드)은 실제로는 **새로운 문자열 객체를 생성**합니다. 이는 메모리 사용과 성능에 영향을 줄 수 있으므로, 대량의 문자열 조작 시에는 `list.join()`과 같은 효율적인 방법을 고려해야 합니다.

```python
s1 = "Hello"
print(f"s1의 메모리 주소 (id): {id(s1)}")

s2 = s1 + " World" # 새로운 문자열 객체 생성
print(f"s2의 메모리 주소 (id): {id(s2)}")
print(f"s1은 변경되지 않음: {s1}")

# 대량의 문자열 연결 성능 비교
import time

num_iterations = 100000 # 10만 번 반복
test_string = "a"

# '+' 연산자를 이용한 연결
start_time = time.time()
result_plus = ""
for _ in range(num_iterations):
    result_plus += test_string
end_time = time.time()
print(f"'+' 연산자 연결 시간: {end_time - start_time:.4f}초")

# list.join()을 이용한 연결
start_time = time.time()
string_list = [test_string for _ in range(num_iterations)]
result_join = "".join(string_list)
end_time = time.time()
print(f"'join()' 메서드 연결 시간: {end_time - start_time:.4f}초")
```

## 3. 타입 확인과 변환 (Type Checking & Casting)

```python
num = 10
text = "Python"
print(f"num의 타입: {type(num)}")
print(f"text의 타입: {type(text)}")

num_str = str(123)
print(f"123을 문자열로: '{num_str}', 타입: {type(num_str)}")

str_int = "456"
int_val = int(str_int)
print(f"'{str_int}'을 정수로: {int_val}, 타입: {type(int_val)}")
```

## 5. 연산자 (Operators)

```python
a = 10
b = 3

print(f"덧셈: {a + b}")
print(f"뺄셈: {a - b}")
print(f"곱셈: {a * b}")
print(f"나눗셈: {a / b}")
print(f"몫: {a // b}")
print(f"나머지: {a % b}")
print(f"거듭제곱: {a ** b}")
```
