## 1. 함수의 기초

```python
def greet_name(name): # name이 매개변수
    """이름을 받아 환영 메시지를 출력하는 함수"""
    print(f"Hello, {name}!")
    
greet_name("Alice") # "Alice"가 인자
```

### 1.3 `return`을 이용한 값 반환

```python
def add(a, b):
    return a + b
    
result = add(5, 3)
print(f"덧셈 결과: {result}")
    
def get_person_info():
    name = "Charlie"
    age = 25
    return name, age
    
person_name, person_age = get_person_info()
print(f"이름: {person_name}, 나이: {person_age}")
```

## 2. 매개변수 고급 활용

```python
def describe_person(name, age, city):
    print(f"이름: {name}, 나이: {age}, 도시: {city}")
    
# 위치 인자와 키워드 인자 혼합 사용
describe_person("Grace", city="Busan", age=30)
```

### 2.5 가변 인자 리스트 (`*args`)

```python
def sum_all(*numbers):
    total = 0
    for num in numbers:
        total += num
    return total
    
print(f"합계: {sum_all(1, 2, 3)}")
print(f"합계: {sum_all(10, 20, 30, 40)}") # Corrected: f-string uses double quotes
```

### 2.6 가변 키워드 인자 (`**kwargs`)

```python
def print_info(**details):
    for key, value in details.items():
        print(f"{key}: {value}")
    
print_info(name="Harry", age=28, city="London")
```

## 5. 람다 표현식 (Lambda Expression)

```python
numbers = [1, 2, 3, 4, 5]

# map과 람다: 각 숫자를 제곱
squared_numbers = list(map(lambda x: x * x, numbers))
print(f"제곱된 숫자: {squared_numbers}")

# filter와 람다: 짝수만 필터링
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(f"짝수: {even_numbers}")
```
